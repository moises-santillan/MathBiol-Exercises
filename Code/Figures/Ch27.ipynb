```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Hill(x, n, K, kmax):
    return kmax * x**n / (x**n + K**n)
```

```python
fig, ax = plt.subplots(ncols=1, nrows=2, figsize=(6, 8))

P = np.linspace(0, 3, 200)
lhs = P

ax[0].plot(P, lhs, label='lhs')

K = 2
kmax = 1
rhs = Hill(P, 3, K, kmax)
ax[0].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = 1
kmax = 1
rhs = Hill(P, 3, K, kmax)
ax[0].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = .5
kmax = 1
rhs = Hill(P, 3, K, kmax)
ax[0].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = .2
kmax = 1
rhs = Hill(P, 3, K, kmax)
ax[0].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

ax[0].set_ylim(0, 1.2)
ax[0].set_xlabel('')
ax[0].tick_params(axis='both', which='major', labelsize=14)

ax[0].legend(fontsize=12)
ax[1].plot(P, lhs, label='lhs')

K = 1
kmax = 1
rhs = Hill(P, 3, K, kmax)
ax[1].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = 1
kmax = 1.5
rhs = Hill(P, 3, K, kmax)
ax[1].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = 1
kmax = 2
rhs = Hill(P, 3, K, kmax)
ax[1].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

K = 1
kmax = 3
rhs = Hill(P, 3, K, kmax)
ax[1].plot(P, rhs, label=rf'rhs, $K={K}, k_m={kmax}$')

ax[1].set_xlabel(r'$P$', fontsize=16)
ax[1].tick_params(axis='both', which='major', labelsize=14)

ax[1].legend(fontsize=12)

fig.tight_layout()

fig.savefig('fig-27.01.png', dpi=300)
```
