```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def transform(u, v):
    arrow_lengths = np.sqrt(u*u + v*v)
    len_adjust_factor = np.log(arrow_lengths + 1) / (arrow_lengths+1e-10)
    return u*len_adjust_factor, v*len_adjust_factor

domain = (0, 1.1)
X = np.linspace(domain[0], domain[1], 16)
Y = np.linspace(domain[0], domain[1], 16)

x, y = np.meshgrid(X, Y)
U = x * (1 - x - 0.5 * y)
V = y * (1 - y - 1.5 * x)

U2, V2 = transform(U, V)

fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(X, Y, U2, V2, color='C0', scale=5)

# Draw the points
ax.scatter([0, 1, 0], [0, 0, 1], color="C1")

# Set the axis limits and labels
ax.set_xlabel(r'$n_1$', fontsize=16)
ax.set_ylabel(r'$n_2$', fontsize=16)
ax.set_title(r'A) $\alpha_{12}>0, \alpha_{21}<0$', fontsize=16)
ax.tick_params(axis='both', which='major', labelsize=14)

plt.savefig("fig-13.01.A.png")
```
