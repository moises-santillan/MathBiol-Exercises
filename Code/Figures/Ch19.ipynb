```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a new figure
fig, ax = plt.subplots(figsize=(6, 6))

# Set the labels for the axes
ax.set_xlabel(r'$x$', fontsize=16)
ax.set_ylabel(r'$y$', fontsize=16)

# Create the x-nullcline
x = np.linspace(-2.2, 2.2, 400)
y_nullcline = x - (x**3) / 3
ax.plot(x, y_nullcline, color='C0', linewidth=2, label=r'$x$-Nullcline')

# Create the y-nullcline
ax.axvline(0, color='C1', linewidth=2, label=r'$y$-Nullcline')

# Mark the fixed point
ax.scatter([0], [0], color='C2', s=50, label='Fixed Point')  # Fixed point

ax.legend(fontsize=14, loc='lower left')
ax.tick_params(axis='both', which='major', labelsize=14)

plt.savefig("fig-19.01.png")
```

```python
# Set up the figure
fig, ax = plt.subplots(figsize=(6, 6))

# Define the domain for x and y
x = np.linspace(-15, 15, 30)
y = np.linspace(-100, 100, 30)

# Create a meshgrid for the vector field
X, Y = np.meshgrid(x, y)

# Define the vector field based on the given dynamics
U = 0.1 * (X - (X**3 / 3) - Y)
V = X / 0.1

# Create the quiver plot
ax.quiver(X, Y, U, V, color='olive', scale=2e3)

# Draw the axes
ax.axvline(0, color='C1', lw=2, label='$y-Nullcline$')

# Plot the nullcline
nullcline_x = np.linspace(-6.7, 6.7, 400)
nullcline_y = nullcline_x - (nullcline_x**3 / 3)
ax.plot(nullcline_x, nullcline_y, color='C0', lw=2, label='x-Nullcline')

# Set the limits and labels
ax.set_title(r'$\mu=0.1$', fontsize=16)
ax.set_xlabel('$x$', fontsize=16)
ax.set_ylabel('$y$', fontsize=16)
ax.grid(False)
ax.tick_params(axis='both', which='major', labelsize=14)

# Show the plot
plt.legend(fontsize=14)

plt.savefig("fig-19.02.A.png")
```

```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

# Locate the Figures directory (tries common relative locations)
candidates = [Path('..') / 'JupyterBook' / 'Figures', Path('..') / '..' / 'JupyterBook' / 'Figures', Path('.') / 'JupyterBook' / 'Figures']
for p in candidates:
    if p.exists():
        BASE = p
        break
else:
    raise FileNotFoundError(f"Could not find JupyterBook/Figures. Checked: {candidates}")

# Filenames from the two previous cells
files = ['fig-19.02.A.png', 'fig-19.02.B.png']
for f in files:
    if not (BASE / f).exists():
        raise FileNotFoundError(f"Missing expected image: {BASE / f}")

# Load images
images = [Image.open(BASE / f).convert('RGBA') for f in files]

# Plot side-by-side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, img in zip(axes, images):
    ax.imshow(img)
    ax.axis('off')

plt.subplots_adjust(wspace=0.02, left=0.01, right=0.99, top=0.99, bottom=0.01)
plt.show()

# Optional: save the combined figure
outpath = BASE / 'fig-19-multipanel.png'
fig.savefig(outpath, dpi=300, bbox_inches='tight', pad_inches=0)
print(f"Saved merged figure to {outpath}")
```
